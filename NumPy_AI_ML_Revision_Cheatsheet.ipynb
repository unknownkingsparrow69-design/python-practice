# NumPy Revision Cheat Sheet for AI/ML

A concise but detailed revision notebook covering the NumPy concepts, functions, and patterns most useful for AI, machine learning, data preprocessing, and neural-network mathematics.

**How to use:** Read the comments beside each operation, edit values, and run each code cell. Focus especially on shapes, axes, broadcasting, vectorization, and matrix multiplication.


```python
import numpy as np  # Import NumPy using its standard alias

print("NumPy version:", np.__version__)  # Check installed NumPy version
```

## 1. Array Basics


```python
a = np.array([1, 2, 3])          # Create a 1D array/vector from a Python list
X = np.array([[1, 2], [3, 4]])   # Create a 2D array/matrix

print(a.ndim)       # Number of dimensions/axes -> 1
print(X.shape)      # Length of each dimension -> (2, 2); vital for ML shape debugging
print(X.size)       # Total number of elements -> 4
print(X.dtype)      # Data type stored in the array
print(X.itemsize)   # Memory used by one element, in bytes
print(X.nbytes)     # Total memory occupied by array data, in bytes
```

## 2. Array Creation


```python
print(np.zeros((2, 3)))         # Create a 2x3 array filled with zeros
print(np.ones((2, 3)))          # Create a 2x3 array filled with ones
print(np.full((2, 3), 7))       # Create a 2x3 array filled with a specified value
print(np.empty((2, 3)))         # Allocate without initializing; existing memory values may appear
print(np.arange(0, 10, 2))      # Values from start to below stop, using a fixed step
print(np.linspace(0, 1, 5))     # Generate exactly 5 evenly spaced values, including endpoints
print(np.eye(3))                # Create a 3x3 identity matrix
print(np.zeros_like(X))         # Zeros with the same shape and dtype as X
print(np.ones_like(X))          # Ones with the same shape and dtype as X
print(np.full_like(X, 5))       # Fill with 5 while preserving X's shape and dtype
```

## 3. Data Types


```python
values = np.array([1, 2, 3])
print(values.dtype)                       # Inspect the current element type
print(values.astype(np.float32))          # Convert to float32; common for ML/DL computations
print(values.astype(np.int32))            # Convert to 32-bit integers
print(values.astype(bool))                # Zero becomes False; nonzero becomes True

X_float = X.astype(np.float32)            # Convert model inputs to floating-point values
print(X_float, X_float.dtype)
```

## 4. Reshaping and Dimensions


```python
a = np.arange(6)                       # Create [0, 1, 2, 3, 4, 5]
print(a.reshape(2, 3))                    # Change to 2 rows x 3 columns without changing values
print(a.reshape(-1, 1))                   # Convert into a column vector; -1 infers row count
print(a.reshape(1, -1))                   # Convert into a row vector; -1 infers column count
print(a.flatten())                        # Flatten to 1D and always return an independent copy
print(a.ravel())                          # Flatten to 1D; returns a view when possible
print(np.expand_dims(a, axis=0).shape)    # Add a new axis at position 0 -> (1, 6)
print(np.expand_dims(a, axis=1).shape)    # Add a new axis at position 1 -> (6, 1)
print(np.squeeze(np.array([[[1, 2, 3]]])).shape)  # Remove axes whose size is 1

M = np.array([[1, 2, 3], [4, 5, 6]])
print(M.T)                                # Transpose a 2D matrix by swapping rows and columns
print(np.transpose(M))                    # General transpose; can reorder axes in N-D arrays
```

## 5. Indexing, Slicing, and Fancy Indexing


```python
a = np.array([10, 20, 30, 40, 50, 60])
print(a[0])               # First element
print(a[-1])              # Last element
print(a[1:4])             # Indexes 1, 2, 3; stop index is excluded
print(a[:3])              # From beginning through index 2
print(a[3:])              # From index 3 through the end
print(a[::2])             # Every second element
print(a[::-1])            # Reverse the array
print(a[[0, 2, 4]])       # Fancy indexing: select multiple specific positions

X = np.arange(1, 13).reshape(3, 4)
print(X[0, 0])            # First row, first column
print(X[0, :])            # All columns from the first row
print(X[:, 0])            # All rows from the first column
print(X[:, 1:3])          # All rows, columns 1 and 2
print(X[[0, 2]])          # Select rows 0 and 2
print(X[:, [0, 2]])       # Select columns 0 and 2
```

## 6. Boolean Filtering


```python
a = np.array([-2, 0, 3, 7, 12])
print(a[a > 5])                     # Keep elements greater than 5
print(a[a != 0])                    # Keep nonzero elements
print(a[(a > 0) & (a < 10)])        # Keep values satisfying both conditions
print(a[(a < 0) | (a > 10)])        # Keep values satisfying either condition
print(a[~(a > 5)])                  # Negate the condition

clean = a.copy()                     # Copy to avoid changing the original
clean[clean < 0] = 0                 # Replace negative values with zero
print(clean)
# Use &, |, and ~ for arrays; Python's and/or do not perform element-wise comparisons.
```

## 7. Element-wise Arithmetic and Math


```python
a = np.array([1.2, 2.5, 4.0])
b = np.array([2.0, 4.0, 8.0])
print(a + b)                  # Element-wise addition
print(a - b)                  # Element-wise subtraction
print(a * b)                  # Element-wise multiplication; not matrix multiplication
print(a / b)                  # Element-wise division
print(a ** 2)                 # Square each value
print(np.sqrt(a))             # Element-wise square root
print(np.exp(a))              # Element-wise e^x; used in sigmoid and softmax
print(np.log(a))              # Element-wise natural logarithm
print(np.abs(-a))             # Element-wise absolute value
print(np.round(a, 1))         # Round values to one decimal place
print(np.floor(a))            # Round down
print(np.ceil(a))             # Round up
print(np.maximum(a, 2))       # Element-wise maximum; useful for ReLU
print(np.minimum(a, 2))       # Element-wise minimum
```

## 8. Aggregation and Statistics


```python
X = np.array([[1, 5, 3], [4, 2, 6]])
print(np.sum(X))              # Sum all elements
print(np.mean(X))             # Arithmetic mean of all elements
print(np.median(X))           # Middle value after sorting
print(np.min(X))              # Smallest value
print(np.max(X))              # Largest value
print(np.std(X))              # Standard deviation: spread around the mean
print(np.var(X))              # Variance: squared standard deviation
print(np.argmin(X))           # Flat index of the minimum value
print(np.argmax(X))           # Flat index of the maximum value
print(np.prod(X))             # Product of all values

probabilities = np.array([0.1, 0.7, 0.2])
print(np.max(probabilities))      # Highest probability value -> 0.7
print(np.argmax(probabilities))   # Predicted class index -> 1
```

## 9. Understanding axis


```python
X = np.array([[1, 2, 3], [4, 5, 6]])  # shape=(2 samples, 3 features)
print(X.mean(axis=0))                    # Collapse rows -> one mean per column/feature
print(X.mean(axis=1))                    # Collapse columns -> one mean per row/sample
print(X.sum(axis=0))                     # Sum each feature across samples
print(X.sum(axis=1))                     # Sum features inside each sample
print(X.mean(axis=0, keepdims=True))     # Keep removed axis -> shape remains (1, 3)

# For a 2D dataset:
# axis=0 -> result per column/feature
# axis=1 -> result per row/sample
```

## 10. Broadcasting


```python
X = np.array([[1, 2, 3], [4, 5, 6]])  # shape (2, 3)
feature_mean = np.array([2, 3, 4])       # shape (3,)
print(X - feature_mean)                  # Subtract feature means from every row automatically
print(X + 5)                             # Scalar broadcasts to every element

# Broadcasting compatibility is checked from right to left.
# Dimensions must be equal or one of them must be 1.
# Common AI use: normalization and bias addition, e.g. X @ W + b.
```

## 11. Matrix and Dot Operations


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(A * B)                   # Element-wise multiplication
print(A @ B)                   # Matrix multiplication; preferred readable notation
print(np.matmul(A, B))         # Function form of matrix multiplication

u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
print(np.dot(u, v))            # Vector dot product -> 1*4 + 2*5 + 3*6

X = np.ones((100, 20))         # 100 samples, 20 input features
W = np.ones((20, 64))          # 20 inputs, 64 output units
b = np.zeros(64)               # One bias per output unit
Z = X @ W + b                  # Dense-layer operation; b broadcasts across all samples
print(Z.shape)                 # Result -> (100, 64)
```

## 12. Linear Algebra


```python
A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])
x = np.array([3.0, 4.0])

print(np.linalg.norm(x))          # Vector magnitude; also used for distance and regularization
print(np.linalg.solve(A, b))      # Solve Ax=b directly; preferred over inv(A) @ b
print(np.linalg.inv(A))           # Matrix inverse; requires a square, invertible matrix
print(np.linalg.det(A))           # Determinant of a square matrix
print(np.linalg.matrix_rank(A))   # Number of linearly independent rows/columns
print(np.linalg.eig(A))           # Eigenvalues and corresponding eigenvectors
print(np.linalg.svd(A))           # Singular Value Decomposition; used in PCA/compression concepts
```

## 13. Combining and Splitting


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(np.concatenate([A, B], axis=0))  # Join along existing row axis; add rows
print(np.concatenate([A, B], axis=1))  # Join along existing column axis; add columns
print(np.vstack([A, B]))                # Stack vertically
print(np.hstack([A, B]))                # Stack horizontally
print(np.stack([A, B], axis=0).shape)   # Create a new axis -> shape (2, 2, 2)

values = np.arange(10)
print(np.array_split(values, 3))        # Split into approximately equal parts; unequal parts allowed
print(np.vsplit(np.arange(16).reshape(4,4), 2))  # Split rows into equal sections
print(np.hsplit(np.arange(16).reshape(4,4), 2))  # Split columns into equal sections
```

## 14. Sorting, Unique Values, and Conditions


```python
a = np.array([4, 2, 4, 1, 3, 2])
print(np.sort(a))                        # Return values sorted in ascending order
print(np.argsort(a))                     # Return indices that would sort the array
print(np.unique(a))                      # Return unique values
classes, counts = np.unique(a, return_counts=True)  # Get values and their frequencies
print(classes, counts)
print(np.where(a > 2, 1, 0))             # Element-wise if-else: True->1, False->0
print(np.clip(a, 2, 3))                  # Force values into inclusive range [2, 3]
```

## 15. Validation and Missing Values


```python
a = np.array([1.0, np.nan, np.inf, -np.inf, 5.0])
print(np.isnan(a))                # True where values are NaN
print(np.isinf(a))                # True where values are positive/negative infinity
print(np.isfinite(a))             # True where values are valid finite numbers
print(np.any(a > 4))              # True if at least one element satisfies the condition
print(np.all(np.array([1,2,3]) > 0))  # True only if every element satisfies the condition
print(np.isclose(0.1 + 0.2, 0.3))     # Safe approximate comparison for floats
print(np.allclose([0.1+0.2], [0.3]))   # Check all corresponding values approximately match
print(np.nanmean(a[np.isfinite(a)]))   # Mean after selecting finite values

clean = a.copy()
clean[np.isnan(clean)] = 0       # Replace NaN with zero
print(clean)
```

## 16. Random Numbers and Reproducibility


```python
rng = np.random.default_rng(42)       # Modern random generator with reproducible seed
print(rng.random((2, 3)))              # Random floats in [0, 1)
print(rng.integers(0, 10, size=5))     # Five integers from 0 through 9
print(rng.normal(0, 1, size=5))        # Samples from normal distribution: mean=0, std=1
print(rng.uniform(0, 1, size=5))       # Samples from uniform distribution
print(rng.choice([10,20,30], size=2))  # Randomly choose values

a = np.arange(5)
rng.shuffle(a)                          # Shuffle in place; modifies a
print(a)
print(rng.permutation(a))               # Return shuffled copy; original is unchanged
```

## 17. Copy vs View


```python
a = np.array([10, 20, 30, 40])
view = a[1:3]                # Slicing commonly returns a view sharing memory with a
view[0] = 999                # This also changes the corresponding element in a
print(a)

independent = a[1:3].copy()  # Create independent data
independent[0] = -1          # Original a is unaffected
print(a, independent)
```

## 18. Additional Statistical and Utility Functions


```python
x = np.array([1, 2, 3, 4, 100])
print(np.percentile(x, 95))          # Value below which 95% of observations fall
print(np.quantile(x, 0.95))          # Same idea expressed using a 0-to-1 fraction
print(np.corrcoef(x, x * 2))         # Correlation matrix; measures linear association
print(np.cov(x, x * 2))              # Covariance matrix
print(np.cumsum([1,2,3]))            # Running sum -> [1,3,6]
print(np.cumprod([1,2,3]))           # Running product -> [1,2,6]
print(np.repeat([1,2], 2))           # Repeat each element -> [1,1,2,2]
print(np.tile([1,2], 2))             # Repeat the whole pattern -> [1,2,1,2]
print(x.tolist())                     # Convert ndarray to a Python list
```

## 19. AI/ML Essential Patterns

These compact patterns combine the NumPy concepts used repeatedly in machine learning.


```python
# Feature standardization / Z-score normalization
X = np.array([[1., 10.], [2., 20.], [3., 30.]])
mean = X.mean(axis=0)                      # Compute one mean for each feature
std = X.std(axis=0)                        # Compute one standard deviation for each feature
X_standardized = (X - mean) / (std + 1e-8)  # Center and scale; epsilon avoids division by zero
print(X_standardized)

# Min-Max scaling
minimum = X.min(axis=0)                    # Minimum of each feature
maximum = X.max(axis=0)                    # Maximum of each feature
X_minmax = (X - minimum) / (maximum - minimum + 1e-8)  # Scale features near [0,1]
print(X_minmax)

# Linear regression-style prediction
W = np.array([0.5, 0.1])                   # One weight per input feature
b = 0.2                                    # Scalar bias
prediction = X @ W + b                     # Weighted sum plus bias
print(prediction)

# Regression losses
y_true = np.array([2., 3., 4.])
y_pred = np.array([2.2, 2.7, 4.1])
mse = np.mean((y_true - y_pred) ** 2)      # MSE: penalizes large errors more strongly
mae = np.mean(np.abs(y_true - y_pred))     # MAE: average absolute prediction error
print("MSE:", mse, "MAE:", mae)
```

## 20. Activation Functions and Numerical Stability


```python
x = np.array([-2., 0., 2.])
relu = np.maximum(0, x)                    # ReLU: negative->0, positive unchanged
sigmoid = 1 / (1 + np.exp(-x))            # Sigmoid: map values into the interval (0,1)
print("ReLU:", relu)
print("Sigmoid:", sigmoid)

logits = np.array([[1000., 1001., 1002.], [1., 2., 3.]])
stable_logits = logits - np.max(logits, axis=1, keepdims=True)  # Prevent exp overflow
exp_scores = np.exp(stable_logits)                              # Convert logits to positive scores
softmax = exp_scores / np.sum(exp_scores, axis=1, keepdims=True) # Normalize rows to sum to 1
print("Softmax:
", softmax)
print("Row sums:", softmax.sum(axis=1))
```

## 21. Distance and Similarity


```python
a = np.array([1., 2., 3.])
b = np.array([2., 4., 6.])
euclidean = np.linalg.norm(a - b)           # Straight-line distance between vectors
cosine = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))  # Directional similarity
print("Euclidean distance:", euclidean)
print("Cosine similarity:", cosine)
```

## 22. Vectorization


```python
a = np.arange(1_000_000)
result = a ** 2        # Vectorized operation over the entire array; avoids a Python element loop
print(result[:5])

# Prefer: result = a ** 2
# Avoid:  [value ** 2 for value in a]
# Vectorized expressions are clearer and normally much faster for numerical workloads.
```

## 23. Shape Pattern to Memorize for AI

For a dense neural-network layer:

- `X`: `(batch_size, input_features)`
- `W`: `(input_features, output_features)`
- `b`: `(output_features,)`
- `Z = X @ W + b`: `(batch_size, output_features)`

This combines shape reasoning, matrix multiplication, broadcasting, vectorization, and activation functions.


```python
batch_size, input_features, output_features = 32, 100, 64
X = np.ones((batch_size, input_features))
W = np.ones((input_features, output_features))
b = np.zeros(output_features)
Z = X @ W + b                 # (32,100) @ (100,64) + (64,) -> (32,64)
A = np.maximum(0, Z)          # Apply ReLU element-wise
print("X:", X.shape, "W:", W.shape, "b:", b.shape, "A:", A.shape)
```
